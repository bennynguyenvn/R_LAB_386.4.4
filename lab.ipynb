## R_GLAB_386.4.4 Aggregations

```python
import pandas as pd
import numpy as np
```

```python
data = {
    'Category': ['A', 'A', 'B', 'B', 'A', 'B','A','B'],
    'Value': [10, 15, 20, 25, 30, 35,40,50],
    'Fee' :[20000,25000,26000,22000,24000,35000,4000,5600],
    'Duration':['30day','40days','35days','40days','60days','60days','50days','60days'],
    'Discount':[1000,2300,1200,2500,2000,2000,500,600]
}

df = pd.DataFrame(data)
print(df.head())
```

```python
# calculate the total sum of he Value column
total_sum = df['Value'].agg('sum')
print(f'Total Sum: {total_sum}')

# ALternative

total_sum1 = df[['Value']].sum()
print(f'Total Sum: {total_sum1}')
```

```python
df['Value'].agg('cumsum')
```

```python
df[['Value']].cumsum()
```

```python
# Do both the agg and the alternative to find: mean of the Value column
mean = df['Value'].agg('mean')
print(f'Mean: {mean}')

# ALternative

mean1 = df[['Value']].mean()
print(f'Mean: {mean1})
```

```python
# Do both the agg and the alternative to find: max of the Value column
max = df['Value'].agg('max')
print(f'Max: {max}')

# ALternative

max1 = df[['Value']].max()
print(f'Max: {max1}')
```

```python
# Do both the agg and the alternative to find: count of the Value column
count = df['Value'].agg('count')
print(f'Max: {count}')

# ALternative

count1 = df[['Value']].count()
print(f'Count: {count1}')
```

## Multiple Aggregations

```python
result = df[['Fee','Discount']].agg('sum')
result
```

```python
result = df[['Fee','Discount','Value']].sum()
result
```

```python
result = df[['Fee','Discount','Value']].agg(['min','max','mean'])
result
```

```python
result_group = df.groupby('Category')[['Value','Fee','Discount']].agg(['sum','mean','max'])
result_group
```
